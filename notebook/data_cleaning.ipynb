```python
import re
def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'', text)
```

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'', text)
```

```python
import string 
exclude = string.punctuation
```

```python
def remove_punc(text):
    for char in exclude:
        text = text.replace(char,'')
    return text
        
```

```python
from nltk.corpus import stopwords
```

```python
def preprocessor(df):
    df["Title"] = df["Title"].str.lower()
    
    df["Title"] = df['Title'].str.replace(r'\s*\([^)]*\)$', '',regex=True)
    
    df["Description"] = df["Description"].apply(lambda x : x.replace("\\"," "))
    df['Description'] = df['Description'].str.replace(r'^[^-]+-\s*', '', regex=True)
    
    
    df["Description"]=df["Description"].apply(remove_html_tags)
    df["Description"]=df["Description"].apply(remove_url)
    df["Description"]=df["Description"].apply(remove_punc)
    
    df["Title"]=df["Title"].apply(remove_html_tags)
    df["Title"]=df["Title"].apply(remove_url)
    df["Title"]=df["Title"].apply(remove_punc)
    stop_words = set(stopwords.words("english"))
    df_train['Description'] = df_train['Description'].str.split().apply(
    lambda words: " ".join(
    word for word in words
    if word.lower() not in stop_words
    )
    )
    
    df_train['Title'] = df_train['Title'].str.split().apply(
    lambda words: " ".join(
    word for word in words
    if word.lower() not in stop_words
    )
    )   
    df["clean"] = df["Title"] + " " + df["Description"]
    
    
    
```

tokenizer 

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vectorizer = TfidfVectorizer(
    stop_words="english", ngram_range=(1, 2),
    max_features=100000, sublinear_tf=True
)

```

```python
x = train["clean"]
y= train["Class Index"]
```

```python
x_vec = vectorizer.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_val, y_train, y_val = train_test_split(
    x_vec, y, test_size=0.1, stratify=y, random_state=42
)
```

# model training :


```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score,classification_report
from sklearn.pipeline import Pipeline
```

```python
nb_model = MultinomialNB()
nb_model.fit(x_train, y_train)

nb_pred = nb_model.predict(x_val)
print("Naive Bayes validation accuracy:", round(accuracy_score(y_val, nb_pred), 4))
```

```python
x_test = test["clean"]
y_test = test["Class Index"]
```

```python
x_test_vec = vectorizer.transform(x_test)
```

```python
y_pred = nb_model.predict(x_test_vec)
```

```python
print("Test Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
svm_model = LinearSVC(C=0.5, random_state=42)
svm_model.fit(x_train, y_train)

svm_pred = svm_model.predict(x_val)
print("Linear SVM validation accuracy:", round(accuracy_score(y_val, svm_pred), 4))
```

```python
y_pred = svm_model.predict(x_test_vec)
```

```python
print("Test Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```
